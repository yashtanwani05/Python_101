```python
sales_data = ["1200", "1500", "1100", "1800", "1600", "1400", "1700"]
sales_numbers= [int(sale) for sale in sales_data]
for  i, sale in enumerate(sales_numbers, start=1):
    print(f"Day {i}: {sale}" )

total_sales = sum(sales_numbers)
average_sales = total_sales / len(sales_numbers)
highest_sales = max(sales_numbers)
lowest_sales = min(sales_numbers)

print("\nTotal Sales: Rs.", total_sales)
print("Average Sales: Rs.", round(average_sales, 2))
print("Highest Sales: Rs.", highest_sales)
print("Lowest Sales: Rs.", lowest_sales)
```
