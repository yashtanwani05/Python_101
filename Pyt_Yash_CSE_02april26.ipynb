```python

raw_data = [" 23.5 ", "low", " 45.0", "HIGH ", "30.2"]

cleaned_data = []
for item in raw_data:
    cleaned_item = item.strip().lower()
    cleaned_data.append(cleaned_item)

numeric_values = []
for item in cleaned_data:
    try:
        number = float(item)
        numeric_values.append(number)
    except ValueError:
        # Ignore non-numeric values
        pass


print("Cleaned Data:", cleaned_data)
print("Numeric Values:", numeric_values)
```

```python
# Function to analyze stock trends
def analyze_trends(stock_data):
    #  Calculate average price
    total = 0
    count = 0
    
    for price in stock_data.values():
        total += price
        count += 1
    
    average = total / count
    
    #  Identify volatility days
    results = []
    
    for date, price in stock_data.items():
        if price >= average * 1.05:
            status = "High Volatility (Above)"
        elif price <= average * 0.95:
            status = "High Volatility (Below)"
        else:
            status = "Stable"
        
        results.append((date, status))
    
    return average, results


# Sample stock data
stock_prices = {
    "2026-04-01": 100,
    "2026-04-02": 108,
    "2026-04-03": 95,
    "2026-04-04": 102,
    "2026-04-05": 89
}

# function
average_price, analysis_result = analyze_trends(stock_prices)

# results
print("Average Price:", average_price)
print("\nTrend Analysis:")

for date, status in analysis_result:
    print(f"{date} -> {status}")
```
